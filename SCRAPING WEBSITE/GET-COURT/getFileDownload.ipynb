```python
import csv
import requests

def download_files_from_csv(csv_file_path, output_folder):
    with open(csv_file_path, 'r', encoding='cp1252') as file:
        csv_reader = csv.reader(file)
        next(csv_reader)  # Skip header row if present
        for row in csv_reader:
            url = row[20]  # assuming the URL is in the first column
            if url != "":
                download_file(url, output_folder)

def download_file(url, output_folder):
    headers = {
        "User-Agent" : (
            "Mozilla/5.0 (X11; linux x86_64)"
            "AppleWebKit/537.36 "
            "(KHTML, like Gecko) Chrome/131.0 Safari/537.36"
        ),
        "Referer": "https://putusan3.mahkamahagung.go.id/",
    }
    response = requests.get(url, headers=headers, timeout=30)
    if response.status_code == 200:
        file_name = url.split("/")[-1]
        file_path = f"{output_folder}/{file_name}.pdf"
        with open(file_path, 'wb') as file:
            file.write(response.content)
        print(f"Downloaded: {file_name}")
    else:
        print(
            f"Failed: status={response.status_code}"
            f"content_type: {response.headers.get('content-type')}, url={url}"
        )

if __name__ == "__main__":
    csv_file_path = "./metaPutusan1.csv"  # Ubah path sesuai dengan lokasi file CSV Anda
    output_folder = "./OUTPUT"  # Ubah path sesuai dengan lokasi folder output Anda
    download_files_from_csv(csv_file_path, output_folder)
```
